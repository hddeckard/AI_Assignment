Prompt: Now can you give me a visualization of the fraud vs non fraud specifically with the transactions with pin vs no pin?

```python
# Set the visual style
sns.set_theme(style="whitegrid")

# Create a grouped count plot with green for non-fraud and red for fraud
plt.figure(figsize=(8, 5))
ax = sns.countplot(
    data=df_cleaned, 
    x='used_pin_number', 
    hue='fraud', 
    palette={0.0: '#2ecc71', 1.0: '#e74c3c'} # Green for Non-Fraud, Red for Fraud
)

# Add title and labels
plt.title('Transactions with PIN vs Fraudulent Transactions', fontsize=14, fontweight='bold')
plt.xlabel('Used PIN', fontsize=11)
plt.ylabel('count', fontsize=11)

# Customize the legend
plt.legend(title='Fraud', labels=['Non-Fraudulent', 'Fraudulent'], loc='upper right')

plt.tight_layout()
plt.show()
```

Prompt: Can you give me a bar graph that uses the percentage of fraud transactions with pin usage? The y axis is the percentage and the x axis is pin vs no pin. The percent that is non fraud should be green and the fraud should be red

AI: Gemini 3

```python
# Create a crosstab of 'used_pin_number' and 'fraud' normalized by index (row percentages)
cross_tab = pd.crosstab(
    df_cleaned['used_pin_number'], 
    df_cleaned['fraud'], 
    normalize='index'
) * 100

# Rename columns for clarity: 0.0 = Non-Fraud, 1.0 = Fraud
cross_tab = cross_tab.rename(columns={0.0: 'Non-Fraudulent', 1.0: 'Fraudulent'})

# Set the visual style
sns.set_theme(style="whitegrid")

# Create a stacked bar plot with specific colors (Green for Non-Fraud, Red for Fraud)
fig, ax = plt.subplots(figsize=(8, 5))
colors = ['#2ecc71', '#e74c3c']  # Green for non-fraud, Red for fraud
cross_tab.plot(kind='bar', stacked=True, color=colors, ax=ax, edgecolor='black', linewidth=0.8)

# Add title and labels
plt.title('Percentage of Fraudulent Transactions with PIN Usage', fontsize=14, fontweight='bold')
plt.xlabel('Used PIN', fontsize=11)
plt.ylabel('Percentage', fontsize=11)

# Rotate x-axis labels for readability and format legend
plt.xticks(ticks=[0, 1], labels=['0.0', '1.0'], rotation=90)
plt.legend(title='Fraud', loc='upper right')

# Set y-axis limits from 0 to 100%
plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```

Prompt: Can you do a scatterplot of fraud vs non fraud for distance from last transaction and ratio to median purchase price?

AI: Gemini 3

```python
# Set the visual style
sns.set_theme(style="whitegrid")

# Create the scatter plot
plt.figure(figsize=(10, 6))
ax = sns.scatterplot(
    data=df_cleaned,
    x='distance_from_last_transaction',
    y='ratio_to_median_purchase_price',
    hue='fraud',
    palette={0.0: '#2ecc71', 1.0: '#e74c3c'}, # Green for Non-Fraud (0), Red for Fraud (1)
    alpha=0.7,
    s=50
)

# Add title and lowercase labels
plt.title('Fraudulent vs Non-Fraudulent Transactions: Distance from Last Transaction vs Ratio to Median Purchase Price', fontsize=12, fontweight='bold')
plt.xlabel('distance_from_last_transaction', fontsize=11)
plt.ylabel('ratio_to_median_purchase_price', fontsize=11)

# Fix the legend handles and labels properly
handles, _ = ax.get_legend_handles_labels()
plt.legend(handles=handles, title='Fraud', labels=['Non-Fraudulent', 'Fraudulent'], loc='upper right')

plt.tight_layout()
plt.show()
```
